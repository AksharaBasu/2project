```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10):
        super(SimpleCNN, self).__init__()
        # Input shape: (Batch, 3, 32, 32)
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        # After two pooling operations, spatial dimensions drop from 32x32 -> 16x16 -> 8x8
        self.fc1 = nn.Linear(64 * 8 * 8, 128)
        self.fc2 = nn.Linear(128, num_classes)
        self.dropout = nn.Dropout(0.25)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(x.size(0), -1)  # Flatten spatial feature map
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x

# Example usage with dummy batch: 4 RGB images of size 32x32
model = SimpleCNN(num_classes=10)
dummy_input = torch.randn(4, 3, 32, 32)
output = model(dummy_input)
print("PyTorch Output Shape:", output.shape)  # Should be [4, 10]
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# 1. Define the CNN Architecture
class CNNModel(nn.Module):
    def __init__(self, num_classes=10):
        super(CNNModel, self).__init__()
        # Input shape: (batch_size, 3, 32, 32)
        self.features = nn.Sequential(
            nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),  # Output: (32, 16, 16)

            nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)   # Output: (64, 8, 8)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 8 * 8, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

# 2. Setup Device, Model, Loss, and Optimizer
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = CNNModel(num_classes=10).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 3. Create Dummy Dataset for Demonstration
# 1000 synthetic RGB images of size 32x32 and labels 0-9
x_data = torch.randn(1000, 3, 32, 32)
y_data = torch.randint(0, 10, (1000,))

dataset = TensorDataset(x_data, y_data)
train_loader = DataLoader(dataset, batch_size=32, shuffle=True)

# 4. Training Loop
epochs = 5
print(f"Training on device: {device}\n")

for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        # Forward pass
        outputs = model(inputs)
        loss = criterion(outputs, labels)

        # Backward pass & optimization
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # Statistics
        running_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

    epoch_loss = running_loss / total
    epoch_acc = (correct / total) * 100
    print(f"Epoch [{epoch+1}/{epochs}] - Loss: {epoch_loss:.4f} | Accuracy: {epoch_acc:.2f}%")
```

```python
import torch
import torchvision.transforms as transforms
from torchvision.models import resnet18, ResNet18_Weights
from PIL import Image
import requests
import io
import gradio as gr

# 1. Load Pre-trained ResNet18 Model & ImageNet Labels
weights = ResNet18_Weights.DEFAULT
model = resnet18(weights=weights)
model.eval()
categories = weights.meta["categories"]

# 2. Image Preprocessing Pipeline
preprocess = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),
])

# 3. Prediction Function from URL
def classify_image_from_url(url):
    try:
        if not url or not url.strip():
            return None, "Please provide a valid image URL."

        # Fetch image from web link
        headers = {'User-Agent': 'Mozilla/5.0'}
        response = requests.get(url.strip(), headers=headers, timeout=10)
        response.raise_for_status()

        image = Image.open(io.BytesIO(response.content)).convert("RGB")

        # Run model inference
        input_tensor = preprocess(image).unsqueeze(0)
        with torch.no_grad():
            output = model(input_tensor)
            probabilities = torch.nn.functional.softmax(output[0], dim=0)

        # Format Top 5 prediction probabilities
        top5_prob, top5_catid = torch.topk(probabilities, 5)
        confidences = {
            categories[top5_catid[i]]: float(top5_prob[i]) for i in range(5)
        }

        return image, confidences

    except Exception as e:
        return None, f"Error processing URL: {str(e)}"

# 4. Build Gradio Web Application UI
with gr.Blocks(title="CNN Image URL Classifier") as demo:
    gr.Markdown("# 🖼️ CNN Image Classifier via Link")
    gr.Markdown("Enter a direct image URL below to fetch the image and get CNN predictions.")

    with gr.Row():
        with gr.Column():
            url_input = gr.Textbox(
                label="Image URL",
                placeholder="https://raw.githubusercontent.com/pytorch/hub/master/images/dog.jpg"
            )
            submit_btn = gr.Button("Run Prediction", variant="primary")

        with gr.Column():
            image_output = gr.Image(label="Fetched Image", type="pil")
            label_output = gr.Label(label="Top Predictions", num_top_classes=5)

    submit_btn.click(
        fn=classify_image_from_url,
        inputs=url_input,
        outputs=[image_output, label_output]
    )

# 5. Launch Local Web Application
if __name__ == "__main__":
    demo.launch()
```

```python

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing import image

from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns


# ============================================================
# 2. DATASET PATH
# ============================================================

# Dataset folder should contain:
#
# dataset/
#    cats/
#       cat1.jpg
#       cat2.jpg
#       ...
#
#    dogs/
#       dog1.jpg
#       dog2.jpg
#       ...

dataset_path = "dataset"


# ============================================================
# 3. IMAGE PREPROCESSING
# ============================================================

IMG_SIZE = 150
BATCH_SIZE = 32

datagen = ImageDataGenerator(
    rescale=1.0 / 255,
    validation_split=0.2
)


# ============================================================
# 4. LOAD TRAINING DATA
# ============================================================

train_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="training",
    shuffle=True
)


# ============================================================
# 5. LOAD VALIDATION DATA
# ============================================================

validation_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="validation",
    shuffle=False
)


# Display class labels
print("Class indices:")
print(train_data.class_indices)


# ============================================================
# 6. BUILD CNN MODEL
# ============================================================

model = Sequential([

    # First Convolution Layer
    Conv2D(
        32,
        (3, 3),
        activation="relu",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    ),

    # First Pooling Layer
    MaxPooling2D(pool_size=(2, 2)),


    # Second Convolution Layer
    Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    # Second Pooling Layer
    MaxPooling2D(pool_size=(2, 2)),


    # Third Convolution Layer
    Conv2D(
        128,
        (3, 3),
        activation="relu"
    ),

    # Third Pooling Layer
    MaxPooling2D(pool_size=(2, 2)),


    # Convert feature maps into one-dimensional vector
    Flatten(),


    # Fully Connected Layer
    Dense(128, activation="relu"),


    # Dropout to reduce overfitting
    Dropout(0.5),


    # Output Layer
    # 0 = Cat
    # 1 = Dog
    Dense(1, activation="sigmoid")
])


# ============================================================
# 7. DISPLAY MODEL ARCHITECTURE
# ============================================================

model.summary()


# ============================================================
# 8. COMPILE THE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 9. TRAIN THE MODEL
# ============================================================

EPOCHS = 10

history = model.fit(
    train_data,
    validation_data=validation_data,
    epochs=EPOCHS
)


# ============================================================
# 10. EVALUATE THE MODEL
# ============================================================

loss, accuracy = model.evaluate(validation_data)

print("\nValidation Loss:", loss)
print("Validation Accuracy:", accuracy)


# ============================================================
# 11. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("CNN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# 12. LOSS GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("CNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# 13. PREDICTIONS
# ============================================================

validation_data.reset()

predictions = model.predict(validation_data)

# Convert probabilities into 0 or 1
predicted_classes = (predictions > 0.5).astype(int).flatten()

actual_classes = validation_data.classes


# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    actual_classes,
    predicted_classes
)

print("\nConfusion Matrix:")
print(cm)


plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Cat", "Dog"],
    yticklabels=["Cat", "Dog"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()


# ============================================================
# 15. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        actual_classes,
        predicted_classes,
        target_names=["Cat", "Dog"]
    )
)


# ============================================================
# 16. PREDICT A NEW IMAGE
# ============================================================

def predict_image(img_path):

    img = image.load_img(
        img_path,
        target_size=(IMG_SIZE, IMG_SIZE)
    )

    img_array = image.img_to_array(img)

    img_array = img_array / 255.0

    img_array = np.expand_dims(
        img_array,
        axis=0
    )

    prediction = model.predict(img_array)[0][0]

    if prediction > 0.5:
        result = "DOG"
        confidence = prediction * 100
    else:
        result = "CAT"
        confidence = (1 - prediction) * 100

    print("Prediction:", result)
    print("Confidence: {:.2f}%".format(confidence))

    plt.imshow(img)
    plt.title(
        "{} ({:.2f}%)".format(result, confidence)
    )
    plt.axis("off")
    plt.show()


# Example:
# predict_image("test/dog.jpg")
# predict_image("test/cat.jpg")


# ============================================================
# 17. SAVE THE TRAINED MODEL
# ============================================================

model.save("cat_dog_cnn_model.keras")

print("\nModel saved successfully!")
```

```python
# ============================================================
# CAT vs DOG CNN - COMPLETE PROJECT
# ============================================================

# -----------------------------
# 1. INSTALL REQUIRED PACKAGE
# -----------------------------
!pip install -q tensorflow-datasets importlib_resources


# -----------------------------
# 2. IMPORT LIBRARIES
# -----------------------------
import tensorflow as tf
import tensorflow_datasets as tfds
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras import models, layers

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    roc_auc_score
)

print("Libraries imported successfully!")


# -----------------------------
# 3. LOAD DATASET
# -----------------------------
(ds_train, ds_validation), ds_info = tfds.load(
    "cats_vs_dogs",
    split=["train[:80%]", "train[80%:]"],
    as_supervised=True,
    with_info=True
)

print("Dataset loaded successfully!")


# -----------------------------
# 4. PREPROCESS DATA
# -----------------------------
IMG_SIZE = 150
BATCH_SIZE = 32

def preprocess(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.cast(image, tf.float32) / 255.0
    return image, label


train_data = (
    ds_train
    .map(preprocess)
    .shuffle(1000)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

validation_data = (
    ds_validation
    .map(preprocess)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

print("Data preprocessing completed!")


# -----------------------------
# 5. CREATE CNN MODEL
# -----------------------------
cnn_model = models.Sequential([

    layers.Input(shape=(150, 150, 3)),

    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),

    layers.Dense(1, activation="sigmoid")
])


# -----------------------------
# 6. COMPILE MODEL
# -----------------------------
cnn_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("\nCNN MODEL SUMMARY")
print("=================")
cnn_model.summary()


# -----------------------------
# 7. TRAIN MODEL
# -----------------------------
print("\nTRAINING MODEL...")
print("=================")

history = cnn_model.fit(
    train_data,
    validation_data=validation_data,
    epochs=10
)


# -----------------------------
# 8. EVALUATE MODEL
# -----------------------------
loss, accuracy = cnn_model.evaluate(
    validation_data,
    verbose=1
)

print("\nMODEL PERFORMANCE")
print("=================")
print("Validation Accuracy:", accuracy * 100, "%")
print("Validation Loss:", loss)


# ============================================================
# 9. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# 10. LOSS GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# 11. GENERATE PREDICTIONS
# ============================================================

y_true = []
y_prob = []

for images, labels in validation_data:

    predictions = cnn_model.predict(
        images,
        verbose=0
    )

    y_true.extend(labels.numpy())
    y_prob.extend(predictions.flatten())


y_true = np.array(y_true)
y_prob = np.array(y_prob)

# Probability >= 0.5 = Dog (1)
# Probability < 0.5 = Cat (0)

y_pred = (y_prob >= 0.5).astype(int)

print("\nPredictions generated successfully!")
print("Number of validation samples:", len(y_true))


# ============================================================
# 12. CLASSIFICATION REPORT
# ============================================================

print("\n")
print("CLASSIFICATION REPORT")
print("=====================")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=["Cat", "Dog"]
    )
)


# ============================================================
# 13. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

print("\nCONFUSION MATRIX")
print("================")
print(cm)


disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Cat", "Dog"]
)

disp.plot()

plt.title("Confusion Matrix - Cat vs Dog")
plt.show()


# ============================================================
# 14. AUC SCORE
# ============================================================

auc_score = roc_auc_score(
    y_true,
    y_prob
)

print("\nAUC SCORE")
print("=========")
print("AUC:", auc_score)


# ============================================================
# 15. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(
    y_true,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label="ROC Curve (AUC = {:.3f})".format(auc_score)
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve - Cat vs Dog CNN")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 16. SAVE MODEL
# ============================================================

model_path = "/content/cat_dog_cnn_model.keras"

cnn_model.save(model_path)

print("\nMODEL SAVED SUCCESSFULLY!")
print("=========================")
print(model_path)


# ============================================================
# 17. DOWNLOAD MODEL
# ============================================================

from google.colab import files

files.download(model_path)

print("\nModel download started!")
```

```python
# ============================================================
# SENTIMENT ANALYSIS USING LSTM
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    roc_auc_score
)

print("Libraries imported successfully!")


# ============================================================
# 2. PARAMETERS
# ============================================================

VOCAB_SIZE = 10000
MAX_LENGTH = 200


# ============================================================
# 3. LOAD IMDb DATASET
# ============================================================

print("\nLoading IMDb dataset...")

(x_train, y_train), (x_test, y_test) = imdb.load_data(
    num_words=VOCAB_SIZE
)

print("Dataset loaded successfully!")

print("Training samples:", len(x_train))
print("Testing samples:", len(x_test))


# ============================================================
# 4. PAD SEQUENCES
# ============================================================

x_train = pad_sequences(
    x_train,
    maxlen=MAX_LENGTH,
    padding="post",
    truncating="post"
)

x_test = pad_sequences(
    x_test,
    maxlen=MAX_LENGTH,
    padding="post",
    truncating="post"
)

print("\nSequence shape:")
print("X Train:", x_train.shape)
print("X Test :", x_test.shape)


# ============================================================
# 5. CREATE LSTM MODEL
# ============================================================

lstm_model = Sequential([

    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=128,
        input_length=MAX_LENGTH
    ),

    LSTM(
        128,
        return_sequences=False
    ),

    Dropout(0.5),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.5),

    Dense(
        1,
        activation="sigmoid"
    )
])


# ============================================================
# 6. COMPILE MODEL
# ============================================================

lstm_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 7. DISPLAY MODEL
# ============================================================

print("\nLSTM MODEL SUMMARY")
print("==================")

lstm_model.summary()


# ============================================================
# 8. TRAIN MODEL
# ============================================================

print("\nTRAINING LSTM MODEL")
print("===================")

history = lstm_model.fit(
    x_train,
    y_train,
    validation_split=0.2,
    epochs=5,
    batch_size=64,
    verbose=1
)


# ============================================================
# 9. EVALUATE MODEL
# ============================================================

loss, accuracy = lstm_model.evaluate(
    x_test,
    y_test,
    verbose=1
)

print("\nMODEL PERFORMANCE")
print("=================")

print("Test Accuracy:", accuracy * 100, "%")
print("Test Loss:", loss)


# ============================================================
# 10. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("LSTM Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 11. LOSS GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 12. GENERATE PREDICTIONS
# ============================================================

print("\nGenerating predictions...")

y_prob = lstm_model.predict(
    x_test,
    verbose=1
).flatten()

y_pred = (y_prob >= 0.5).astype(int)

y_true = np.array(y_test)

print("Predictions generated successfully!")


# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\n")
print("CLASSIFICATION REPORT")
print("=====================")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=["Negative", "Positive"]
    )
)


# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

print("\nCONFUSION MATRIX")
print("================")

print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negative", "Positive"]
)

disp.plot()

plt.title("LSTM Confusion Matrix")

plt.show()


# ============================================================
# 15. AUC SCORE
# ============================================================

auc_score = roc_auc_score(
    y_true,
    y_prob
)

print("\nAUC SCORE")
print("=========")

print("AUC:", auc_score)


# ============================================================
# 16. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(
    y_true,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label="LSTM ROC Curve (AUC = {:.3f})".format(auc_score)
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve - LSTM Sentiment Analysis")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 17. SAVE MODEL
# ============================================================

model_path = "/content/lstm_sentiment_model.keras"

lstm_model.save(
    model_path
)

print("\nMODEL SAVED SUCCESSFULLY!")
print("=========================")

print(model_path)


# ============================================================
# 18. DOWNLOAD MODEL
# ============================================================

from google.colab import files

files.download(
    model_path
)

print("\nLSTM model download started!")
```

```python
# ============================================================
# SENTIMENT ANALYSIS USING GRU
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense, Dropout

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    roc_auc_score
)

print("Libraries imported successfully!")


# ============================================================
# 2. PARAMETERS
# ============================================================

VOCAB_SIZE = 10000
MAX_LENGTH = 200


# ============================================================
# 3. LOAD IMDb DATASET
# ============================================================

print("\nLoading IMDb dataset...")

(x_train, y_train), (x_test, y_test) = imdb.load_data(
    num_words=VOCAB_SIZE
)

print("Dataset loaded successfully!")

print("Training samples:", len(x_train))
print("Testing samples:", len(x_test))


# ============================================================
# 4. PAD SEQUENCES
# ============================================================

x_train = pad_sequences(
    x_train,
    maxlen=MAX_LENGTH,
    padding="post",
    truncating="post"
)

x_test = pad_sequences(
    x_test,
    maxlen=MAX_LENGTH,
    padding="post",
    truncating="post"
)

print("\nSequence shape:")
print("X Train:", x_train.shape)
print("X Test :", x_test.shape)


# ============================================================
# 5. CREATE GRU MODEL
# ============================================================

gru_model = Sequential([

    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=128,
        input_length=MAX_LENGTH
    ),

    GRU(
        128,
        return_sequences=False
    ),

    Dropout(0.5),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.5),

    Dense(
        1,
        activation="sigmoid"
    )
])


# ============================================================
# 6. COMPILE MODEL
# ============================================================

gru_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 7. DISPLAY MODEL
# ============================================================

print("\nGRU MODEL SUMMARY")
print("=================")

gru_model.summary()


# ============================================================
# 8. TRAIN MODEL
# ============================================================

print("\nTRAINING GRU MODEL")
print("==================")

history = gru_model.fit(
    x_train,
    y_train,
    validation_split=0.2,
    epochs=5,
    batch_size=64,
    verbose=1
)


# ============================================================
# 9. EVALUATE MODEL
# ============================================================

loss, accuracy = gru_model.evaluate(
    x_test,
    y_test,
    verbose=1
)

print("\nMODEL PERFORMANCE")
print("=================")

print("Test Accuracy:", accuracy * 100, "%")
print("Test Loss:", loss)


# ============================================================
# 10. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("GRU Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 11. LOSS GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("GRU Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 12. GENERATE PREDICTIONS
# ============================================================

print("\nGenerating predictions...")

y_prob = gru_model.predict(
    x_test,
    verbose=1
).flatten()

y_pred = (y_prob >= 0.5).astype(int)

y_true = np.array(y_test)

print("Predictions generated successfully!")


# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\n")
print("CLASSIFICATION REPORT")
print("=====================")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=["Negative", "Positive"]
    )
)


# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

print("\nCONFUSION MATRIX")
print("================")

print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negative", "Positive"]
)

disp.plot()

plt.title("GRU Confusion Matrix")

plt.show()


# ============================================================
# 15. AUC SCORE
# ============================================================

auc_score = roc_auc_score(
    y_true,
    y_prob
)

print("\nAUC SCORE")
print("=========")

print("AUC:", auc_score)


# ============================================================
# 16. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(
    y_true,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label="GRU ROC Curve (AUC = {:.3f})".format(auc_score)
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve - GRU Sentiment Analysis")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 17. SAVE MODEL
# ============================================================

model_path = "/content/gru_sentiment_model.keras"

gru_model.save(
    model_path
)

print("\nMODEL SAVED SUCCESSFULLY!")
print("=========================")

print(model_path)


# ============================================================
# 18. DOWNLOAD MODEL
# ============================================================

from google.colab import files

files.download(
    model_path
)

print("\nGRU model download started!")
```
